缓存系统与上游模块进行协同验证十分复杂，会妨碍快速迭代。

Co-verification of the Cache system with upstream modules is complex and prevents rapid iteration.

为此我们开发 TL-Test：一个单元级缓存系统验证框架，支持 TileLink 协议、缓存一致性检查和随机测试用例生成。

To address this issue, we developed TL-Test: a unit-level cache-system verification framework that supports the TileLink protocol, cache-coherence checking, and randomized test-case generation.

下面举例说明 TL-Test 如何检测缓存一致性违规。

Here is another example to detect cache coherence violation by TL-Test.

我们注入一个错误，使 grant 数据发生错误移位。

We inject a bug that wrongly shift the grant data.

TL-Test 生成随机测试并定位缓存设计中特定地址上的传输问题，记录总线事务，再用 `grep` 提取日志分析。

TL-Test generates randomized tests and pinpoints a transfer problem at a specific address in our cache design.

It logs all bus transactions; we then use `grep` to extract log for analysis.
